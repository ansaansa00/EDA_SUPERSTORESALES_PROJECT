## What is Superstore Order Data?

The Superstore dataset contains transaction-level retail order information collected from customers across different regions and product categories. 
The dataset provides insights into sales performance, customer purchasing behavior, product demand, shipping methods, discounts, and profitability.

Each row in the dataset represents a single customer order along with customer details, product information, sales amount, shipping details, and profit generated from the transaction.

Key variables tracked in the dataset include:

- Row ID — Unique identifier for each row in the dataset  
- Order ID — Unique identifier for each customer order  
- Order Date — Date on which the order was placed  
- Ship Date — Date on which the order was shipped  
- Ship Mode — Shipping method used for delivery  
- Customer ID — Unique identifier for each customer  
- Customer Name — Name of the customer  
- Segment — Customer segment such as Consumer, Corporate, or Home Office  
- Country/Region — Country or region of the customer  
- City — City where the order was placed  
- State/Province — State or province of the customer  
- Postal Code — Postal code of the customer location  
- Region — Sales region such as East, West, Central, or South  
- Product ID — Unique identifier for each product  
- Category — Main product category  
- Sub-Category — Subdivision within the product category  
- Product Name — Name of the product  
- Sales — Total sales amount generated from the order  
- Quantity — Number of products purchased  
- Discount — Discount applied to the product  
- Profit — Profit earned from the transaction

## Question 1: Which category generates the highest sales?

```python
category_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(7,5))
category_sales.plot(kind='bar')

plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

The analysis shows that different product categories contribute differently to overall sales. 
The category with the highest sales indicates strong customer demand and higher market performance. 
This insight helps businesses focus more on high-performing categories to improve revenue generation and inventory planning.

## Question 2: Which region contributes the highest sales?

```python
region_sales = df.groupby('Region')['Sales'].sum()

plt.figure(figsize=(6,6))
plt.pie(
    region_sales.values,
    labels=region_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Regional Sales Distribution")
plt.show()
```

The regional sales analysis highlights the variation in sales performance across different regions. 
Regions with higher sales indicate stronger customer demand and better market reach. 
This information can help businesses identify high-performing markets and develop region-specific marketing and sales strategies.

## Question 3: Which customer segment contributes the most to sales?

```python
segment_sales = df.groupby('Segment')['Sales'].sum()

plt.figure(figsize=(7,5))
segment_sales.plot(kind='bar')

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.show()
```

The analysis reveals that customer segments contribute differently to total sales revenue. 
The segment with the highest sales represents the most valuable customer group for the business. 
Understanding customer segment behavior helps organizations improve customer targeting, retention strategies, and personalized marketing.

## Question 4: How are discounts distributed across orders?

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Discount'], bins=20)

plt.title("Distribution of Discounts")
plt.xlabel("Discount")
plt.ylabel("Frequency")

plt.show()
```


The graph shows how discounts are distributed among customer orders. 
Most orders have low or moderate discounts, while only a few orders receive very high discounts. 
This analysis helps businesses understand discount usage patterns and maintain balanced pricing strategies.

## Question 5: Which sub-category has the highest sales?

```python
subcategory_sales = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(12,5))
subcategory_sales.plot(kind='bar')

plt.title("Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

The sub-category analysis identifies products that contribute the most to total sales. 
High-performing sub-categories indicate popular products with strong customer demand. 
This helps businesses improve inventory management, product promotion, and sales planning.

## Question 6: Which region generates the highest profit?

```python
region_profit = df.groupby('Region')['Profit'].sum()

plt.figure(figsize=(7,5))
region_profit.plot(kind='bar')

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.show()
```

Profit distribution varies across regions due to differences in sales performance, operational costs, and discount strategies. 
Regions generating higher profit contribute more effectively to business growth and sustainability. 
This insight supports better regional business planning and resource allocation.

## Question 7: What is the monthly sales trend?

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Month'] = df['Order Date'].dt.month
monthly_sales = df.groupby('Month')['Sales'].sum()
plt.figure(figsize=(8,5))
monthly_sales.plot(kind='line', marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

The .dt accessor works only with datetime data types. 
Since the 'Order Date' column was stored as a string object, it needed to be converted into datetime format before extracting the month.
